# 03 — Chat compare (optional notebook)

Same as `python lessons/03_chat.py`, with a markdown table.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from IPython.display import Markdown, display
from common import chat, choice_text, finish_reason, usage

PROMPT = "Python code to give a sqrt of pi to 6 decimal places."
SYSTEM = "Be terse. Output only what was asked. No greetings or extras."
STOP = ["**Output", "Output:", "Here are", "Sure,"]

base = chat([{"role": "user", "content": PROMPT}], temperature=0.7, max_tokens=800)
cons = chat(
    [
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": PROMPT + "\nReply with only the Python code. No markdown fences."},
    ],
    stop=STOP,
    temperature=0.2,
    max_tokens=400,
)

def row(name, data):
    t = choice_text(data)
    u = usage(data) or {}
    return (
        f"| {name} | {finish_reason(data)} | {u.get('completion_tokens')} | {len(t)} |"
        f" `{t[:60].replace(chr(10), ' ')}` |"
    )

display(Markdown(
    "| run | finish | completion_tokens | chars | preview |\n"
    "|---|---|---:|---:|---|\n"
    + row("baseline", base) + "\n"
    + row("constrained", cons)
))
print("BASELINE\n", choice_text(base))
print("\nCONSTRAINED\n", choice_text(cons))
```
